# Un scorecard con German Credit

Ejemplo mínimo de bayesrisk. Instala el entorno indicado en la guía y extrae la descarga completa. Los datos ya preparados conservan el índice original y excluyen la evaluación exterior agrupada. Autor: Hans Hofmann; UCI, CC BY 4.0. El target original 1/2 se transforma a 0/1, con 1 como malo.

Esta corrida usa únicamente el complemento de la evaluación; el Holdout interno es diagnóstico. Ningún resultado elige el modelo automáticamente. El anexo separado contiene el benchmark completo y sus límites.

```python
from bayesrisk import Scorecard

sc = Scorecard(
    "prepared/german/input.parquet",
    target="target",
    partition="random",
    name="german",
    run_dir="../h10-work/notebook-runs",
    formats=["html", "docx"],
)
sc.run(raise_on_error=True)
```
