# Primera provisión interna

TL;DR: recorrido experimental con PD y LGD declaradas. La PD constante es ilustrativa, no estimada ni validada. Se conserva la fórmula grupal y el detalle es un prorrateo.

Requiere `bayesrisk[scoring,excel]`. Los datos sintéticos F5 vienen con el paquete. Ejecuta las celdas en orden en una carpeta propia; allí se guardan la evidencia y los entregables.

## Calcular y mirar el resumen

```python
from pathlib import Path

import pandas as pd

from bayesrisk import Provision
from bayesrisk.ui.datasets import materialize

cartera = pd.read_parquet(materialize("provision_interna_generica", workdir=Path("datos")))
cartera["pd_declarada"] = 0.02  # Sólo ilustración: no es una PD estimada.
provision = Provision(cartera, as_of="as_of_date", portfolio="portfolio",
                      exposure="exposure_amount", pd="pd_declarada", lgd="lgd", formats=["csv"])
provision.run(until="provisioning_internal")
provision.summary()
```

## Reanudar y exportar

Reanudar crea una corrida completa nueva. El CSV contiene todo el detalle, Excel reúne las etapas y el ZIP conserva entradas y estudios. Sin ID declarado sólo se verifica el índice técnico; la unicidad económica no está acreditada.

```python
provision.resume()
provision.export_excel()
provision.export("provision.zip")
provision.summary("provisioning_internal")
```
